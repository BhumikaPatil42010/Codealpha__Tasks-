# **KARGS**

kwargs lets a function accept any number of named (key=value) arguments.

It collects them into a dictionary.

```python
def show_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} = {value}")

show_details(name="bhumika", age=25, city="Mumbai")
```

```python
# Passing a dictionary as kwargs
def print_info(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)

# Create a dictionary
data = {
    "name": "Anita",
    "age": 22,
    "city": "Mumbai"
}

# Pass dictionary using **
print_info(**data)
```

```python
def display(**kwargs):
  print(kwargs)
display(name ="Bhumika", age=21, city="pune")
```

```python
# factorial of number using function
def factorial(n):
    result = 1
    for i in range(1, n+1):
        result *= i
    return result

print(factorial(5))
```

```python
# using recursion
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)

print(factorial(5))
```

```python
# find the second highest number in a list. without using inbuilt function
def second_highest(numbers):
    largest = numbers[0]
    second = numbers[0]

    for num in numbers:
        if num > largest:
            second = largest
            largest = num
        elif num > second and num != largest:
            second = num

    return second


numbers = [10, 5, 20, 8, 15]
print("Second highest:", second_highest(numbers))
```

```python
#  with built in
def fibonacci_series(n):
    series = [0, 1]
    for i in range(2, n):
        series.append(series[-1] + series[-2])
    return series[:n]

n = int(input("Enter number of terms: "))
print("Fibonacci series:", fibonacci_series(n))
```

```python
# find the vowels count in a string by user input
# Count vowels in a string
def count_vowels(string):
    vowels = "aeiouAEIOU"
    count = 0
    for ch in string:
        if ch in vowels:
            count += 1
    return count

# Driver code
user_input = input("Enter a string: ")
print("Number of vowels in the string:", count_vowels(user_input))

```

```python
# convert the string into lower string into upper string
def convert_case(string):
    lower_str = string.lower()
    upper_str = string.upper()
    return lower_str, upper_str

# Driver code
user_input = input("Enter a string: ")
lower, upper = convert_case(user_input)

print("Lowercase string:", lower)
print("Uppercase string:", upper)
```

```python
# ATM Machine Simulation

BALANCE = 5000   # Default balance
PASSWORD = "1234"  # Set user password

def atm_withdraw():
    global BALANCE  # Use global variable to update balance

    user_password = input("Enter your ATM password: ")

    if user_password == PASSWORD:
        print("Password matched ")

        withdraw = int(input("Enter amount to withdraw: "))

        # Check withdrawal conditions
        if withdraw > BALANCE:
            print("Insufficient Balance")
        elif withdraw < 100:
            print("Cannot withdraw less than 100 Rs")
        elif withdraw % 100 != 0:
            print("Can only withdraw in multiples of 100 Rs (100, 500, etc.)")
        else:
            BALANCE -= withdraw
            print(f"Withdrawal successful! ")
            print(f"Remaining Balance: {BALANCE} Rs")
    else:
        print("Incorrect Password ")

# Call the function
atm_withdraw()
```

```python
# BREAK -  Stops the loop completely when the condition is met.
for i in range(1, 6):
    if i == 3:
        break
    print(i)



```

```python
# CONTINUE - Skips the current iteration and moves to the next loop cycle.
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```
